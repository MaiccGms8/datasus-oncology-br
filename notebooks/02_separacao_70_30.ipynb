```python
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
```

```python
dados = pd.read_csv("../data/oncological_panel_limpo.csv")

print("Base carregada com sucesso!")
```

```python
print("Quantidade de registros:", len(dados))
print("Quantidade de colunas:", len(dados.columns))
```

```python
dados_treino, dados_teste = train_test_split(
    dados,
    test_size=0.30,
    random_state=42
)

print("Separação realizada com sucesso!")
```

```python
print("Total de registros:", len(dados))
print("Registros para treino:", len(dados_treino))
print("Registros para teste:", len(dados_teste))

print()
print("Percentual de treino:",
      round(len(dados_treino) / len(dados) * 100, 2), "%")

print("Percentual de teste:",
      round(len(dados_teste) / len(dados) * 100, 2), "%")
```

```python
dados_treino.to_csv(
    "../data/oncological_panel_treino.csv",
    index=False
)

dados_teste.to_csv(
    "../data/oncological_panel_teste.csv",
    index=False
)

print("Arquivos salvos com sucesso!")
```
